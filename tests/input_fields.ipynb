# Examining Input Field Structure
To see how to make a valid .h5 file for external fields, I am examining an example file from a magnetic mirror tutorial (https://blast-warpx.github.io/warpx-tutorials/a-magnetic-mirror.html)

```python
# path to the input file
example_input_path = BEXT_DIR / 'example-femm-3d.h5'

# open file
f = h5py.File(example_input_path, 'r')
```

```python
def print_tree(name, obj):
    indent = "  " * name.count("/")
    if isinstance(obj, h5py.Group):
        print(f"{indent}{name}/")
    else:
        print(f"{indent}{name} {obj.shape} {obj.dtype}")

# print group and dataset names
f.visititems(print_tree)
```

```python
# close the file (last code block)
f.close()
```

The group structure is as follows:
- data, 1, meshes

meshes contains two groups:
- B
- E

The B and E groups have datasets (one per position component)
- x
- y
- z
